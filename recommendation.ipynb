```python
import pandas as pd
```

```python
movies = pd.read_csv("data/movies_converted.csv")
```

```python
movies.head()
```

```python
movies = movies[["id", "title", "overview", "genres"]]
```

```python
movies.head()
```

```python
movies.loc[:, "tags"] = movies["overview"] + ' ' + movies["genres"] 
```

```python
movies
```

```python
new_data = movies.drop(columns=["overview", "genres"])
```

```python
new_data
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(stop_words="english")
```

```python
cv
```

```python
vector = cv.fit_transform(new_data["tags"].values.astype("U"))
```

```python
vector = vector.toarray()
```

```python
vector.shape
```

```python
vector
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

```python
similarity.shape
```

```python
def recommend(movies):
    index = new_data[new_data["title"]==movies].index[0]
    distance = sorted(list(enumerate(similarity[index])), reverse=True, key=lambda vector:vector[1])
    for i in distance[0:5]:
        print(new_data.iloc[i[0]].title)
```

```python
recommend("The Scorpion King 3: Battle for Redemption")
```
